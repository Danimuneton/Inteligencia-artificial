## 1. Imports y funciones de Q-Learning


```python
import numpy as np
import gymnasium as gym
import random
import matplotlib.pyplot as plt
import time
from IPython.display import clear_output, display

from IPython.display import HTML
from matplotlib import animation
import matplotlib.pyplot as plt


def play_episode(env, Q=None, random_policy=False, max_steps=100, seed=None):
    """Ejecuta un episodio y devuelve sus frames y recompensa total."""
    state, _ = env.reset(seed=seed)
    frames = [env.render()]
    total_reward = 0.0

    for _ in range(max_steps):
        if random_policy:
            action = env.action_space.sample()
        else:
            q_values = Q[state]
            max_q = np.max(q_values)

            # Desempate aleatorio entre acciones con el mismo Q.
            best_actions = np.flatnonzero(q_values == max_q)
            action = int(np.random.choice(best_actions))

        next_state, reward, terminated, truncated, _ = env.step(action)

        frames.append(env.render())
        total_reward += reward
        state = next_state

        if terminated or truncated:
            break

    return frames, total_reward


def frames_to_video(frames, interval=700):
    """Convierte una lista de frames RGB en una animación reproducible en Jupyter."""
    fig = plt.figure(figsize=(4, 4))
    plt.axis("off")

    image = plt.imshow(frames[0])

    def update(frame):
        image.set_data(frame)
        return [image]

    anim = animation.FuncAnimation(
        fig,
        update,
        frames=frames,
        interval=interval,
        blit=True,
        repeat=True
    )

    plt.close(fig)
    return HTML(anim.to_jshtml())


def initialize_q_table(state_space, action_space):
    return np.zeros((state_space, action_space))


def greedy_policy(Qtable, state):
    # Si hay empate entre varias acciones con el mismo Q, desempata al azar.
    max_q = np.max(Qtable[state])
    best_actions = np.flatnonzero(Qtable[state] == max_q)
    return int(np.random.choice(best_actions))


def epsilon_greedy_policy(Qtable, state, epsilon, env):
    if random.random() < epsilon:
        return env.action_space.sample()

    return greedy_policy(Qtable, state)


def train_q_learning(
    env,
    Qtable,
    n_episodes=5000,
    learning_rate=0.7,
    gamma=0.95,
    max_epsilon=1.0,
    min_epsilon=0.05,
    decay_rate=0.001,
    max_steps=100,
    start_episode=0,
):
    rewards = []

    for episode in range(n_episodes):
        state, _ = env.reset()
        total_reward = 0

        global_episode = start_episode + episode
        epsilon = min_epsilon + (
            max_epsilon - min_epsilon
        ) * np.exp(-decay_rate * global_episode)

        for _ in range(max_steps):
            action = epsilon_greedy_policy(
                Qtable, state, epsilon, env
            )

            next_state, reward, terminated, truncated, _ = env.step(action)
            done = terminated or truncated

            best_next_q = 0.0 if done else np.max(Qtable[next_state])

            td_target = reward + gamma * best_next_q
            td_error = td_target - Qtable[state, action]

            Qtable[state, action] += learning_rate * td_error

            state = next_state
            total_reward += reward

            if done:
                break

        rewards.append(total_reward)

    return Qtable, rewards


def evaluate_q_policy(env, Qtable, n_episodes=100, max_steps=100):
    episode_rewards = []

    for _ in range(n_episodes):
        state, _ = env.reset()
        total_reward = 0

        for _ in range(max_steps):
            action = greedy_policy(Qtable, state)

            next_state, reward, terminated, truncated, _ = env.step(action)

            state = next_state
            total_reward += reward

            if terminated or truncated:
                break

        episode_rewards.append(total_reward)

    return np.mean(episode_rewards), np.std(episode_rewards)


def show_frame(env, title=''):
    frame = env.render()
    plt.figure(figsize=(4, 4))
    plt.imshow(frame)
    plt.axis('off')
    plt.title(title)
    display(plt.gcf())
    plt.close()
```

### 💬 Antes de ejecutar

En CheeseWorld teníamos explícitamente una clase `Environment` y una clase `QLearningAgent`.

**Pregunta:** en este notebook, ¿qué papel cumple Gymnasium y dónde quedó representado el agente?

**Respuesta / notas:**

- **Gymnasium es el *ambiente*** (el equivalente a `CheeseWorld`): define los
  estados, las acciones, las transiciones y las recompensas, y responde a
  `env.reset()` y `env.step(action)` devolviendo `(next_state, reward,
  terminated, truncated, info)`. No aprende ni decide.
- **El agente ya no es una clase**, quedó repartido en las *funciones* de
  Q-Learning: la Q-table (`Qtable`), la política `epsilon_greedy_policy`/
  `greedy_policy` (qué acción tomar) y `train_q_learning` (cómo actualizar
  $Q$). Esos tres elementos juntos cumplen el papel que antes tenía
  `QLearningAgent`.
- La separación **ambiente ↔ agente** se mantiene: Gymnasium provee la
  dinámica del mundo; nuestras funciones aprenden sobre él.


## 2. Crear FrozenLake


```python
env = gym.make(
    "FrozenLake-v1",
    map_name="4x4",
    is_slippery=False,
    render_mode="rgb_array"
)

state, info = env.reset()

print("Estado inicial:", state)
print("Número de estados:", env.observation_space.n)
print("Número de acciones:", env.action_space.n)
```

En FrozenLake las acciones son:

| Acción | Código |
|---|---:|
| Left | 0 |
| Down | 1 |
| Right | 2 |
| Up | 3 |

Primero trabajaremos con `is_slippery=False`, es decir, con transiciones determinísticas.


### 💬 Actividad 1 — La Q-table

**Respuesta / discusión:**

- **Filas:** 16, una por cada estado (grid 4×4 ⇒ `observation_space.n = 16`).
- **Columnas:** 4, una por cada acción (Left, Down, Right, Up ⇒
  `action_space.n = 4`).
- **Qué representa $Q[s,a]$:** la estimación del *retorno esperado*
  (recompensa acumulada y descontada) si en el estado $s$ tomo la acción $a$ y
  después sigo la política aprendida. No es la recompensa inmediata, sino el
  valor a largo plazo de ese par estado-acción.


```python
state_space = env.observation_space.n
action_space = env.action_space.n

Q = initialize_q_table(state_space, action_space)

print("Q-table shape:", Q.shape)
Q
```

### 💬 Actividad 2 — Inicio del aprendizaje

Todos los valores son cero.

$$
Q(s,a)=0
$$

**Notas:**

- Que $Q(s,a)=0$ **no** significa que todas las acciones sean malas: significa
  que el agente **todavía no sabe nada**. Es una inicialización neutra, no un
  juicio sobre las acciones.
- Si varias acciones tienen el **mismo valor máximo** (al inicio, las cuatro
  valen 0), el `argmax` quedaría empatado. Por eso `greedy_policy` **desempata
  al azar** (`np.random.choice` entre las acciones con Q máximo); de lo
  contrario el agente elegiría siempre la primera acción y exploraría peor.


## 3. Ejecutar una transición


```python
state, _ = env.reset()

epsilon = 1.0
action = epsilon_greedy_policy(Q, state, epsilon, env)

next_state, reward, terminated, truncated, _ = env.step(action)

print("state      =", state)
print("action     =", action)
print("reward     =", reward)
print("next_state =", next_state)
print("done       =", terminated or truncated)
```

### 💬 Actividad 3 — Identificar la experiencia

La transición anterior se escribe como la experiencia:

$$
(s,\;a,\;r,\;s') = (\texttt{state},\;\texttt{action},\;\texttt{reward},\;\texttt{next\_state})
$$

(los valores concretos son los que imprimió la celda anterior).

**¿De cuál elemento no disponíamos en Value Iteration?**
De la **recompensa $r$ y del siguiente estado $s'$ como *muestras reales***.
En Value Iteration conocíamos el modelo completo $T(s,a,s')$ y $R(s)$ y
podíamos *calcular* el valor esperado sobre todos los $s'$. Aquí, en cambio,
Q-Learning es **model-free**: no conoce $T$ ni $R$; solo observa un $(r, s')$
*muestreado* al interactuar con el ambiente. Esa es la diferencia clave entre
planear con un modelo y aprender por experiencia.


## 4. Del azar a una política aprendida

Vamos a observar **el mismo agente en tres momentos**. Primero no sabe nada y actúa al azar; luego veremos su política después de pocas experiencias; finalmente veremos la política después del entrenamiento completo.


```python
# Guardaremos tres momentos del aprendizaje

# Momento 1: sin entrenamiento
Q_initial = initialize_q_table(
    env.observation_space.n,
    env.action_space.n
)

# Momento 2: poco entrenamiento
EARLY_EPISODES = 50
Q_early = Q_initial.copy()
Q_early, rewards_early = train_q_learning(
    env,
    Q_early,
    n_episodes=EARLY_EPISODES,
    learning_rate=0.7,
    gamma=0.95,
    max_epsilon=1.0,
    min_epsilon=0.05,
    decay_rate=0.001,
)

# Momento 3: continuar hasta 10 000 episodios
TOTAL_EPISODES = 10000
Q_trained = Q_early.copy()
Q_trained, rewards_final = train_q_learning(
    env,
    Q_trained,
    n_episodes=TOTAL_EPISODES - EARLY_EPISODES,
    learning_rate=0.7,
    gamma=0.95,
    max_epsilon=1.0,
    min_epsilon=0.05,
    decay_rate=0.001,
    start_episode=EARLY_EPISODES,
)

Q = Q_trained
rewards = rewards_early + rewards_final

print('Snapshots guardados:')
print('Q_initial : 0 episodios')
print(f'Q_early   : {EARLY_EPISODES} episodios')
print(f'Q_trained : {TOTAL_EPISODES} episodios')
```

### Momento 1 — Sin entrenamiento: random walk
Todavía no usamos la Q-table para decidir. Cada acción se selecciona aleatoriamente. Observa cómo interactúa el agente con el mundo.

```python
frames_random, reward_random = play_episode(
    env,
    Q_initial,
    random_policy=True,
    seed=7
)

print(f"Recompensa total: {reward_random}")
frames_to_video(frames_random, interval=700)
```

### Momento 2 — Después de pocas iteraciones

Ahora el agente usa de forma **greedy** lo que ha aprendido en `Q_early`. Todavía conoce poco del ambiente, así que su comportamiento puede ser incompleto o equivocarse.


```python
frames_early, reward_early = play_episode(
    env,
    Q_early,
    random_policy=False,
    seed=7
)

print(f"Recompensa total: {reward_early}")
frames_to_video(frames_early, interval=700)
```

### Momento 3 — Agente entrenado

Finalmente usamos `Q_trained`. Ya no exploramos: en cada estado el agente selecciona una de las acciones con mayor valor $Q(s,a)$.


```python
frames_trained, reward_trained = play_episode(
    env,
    Q_trained,
    random_policy=False,
    seed=7
)

print(f"Recompensa total: {reward_trained}")
frames_to_video(frames_trained, interval=700)
```

### 💬 Actividad — ¿Qué cambió?

**Respuesta.** El ambiente, los estados y las acciones son idénticos en los
tres momentos. Lo único que cambió es la **Q-table**:

- `Q_initial`: todo en cero ⇒ el agente actúa al azar y rara vez llega a la
  meta.
- `Q_early` (50 episodios): algunos valores empiezan a ser positivos cerca de
  la meta, pero la política aún es incompleta y puede equivocarse.
- `Q_trained` (10000 episodios): los valores $Q(s,a)$ a lo largo del camino
  óptimo son claramente mayores, de modo que la política greedy conduce
  directo del inicio a la meta.

La mejora de comportamiento = valores $Q$ mejor estimados; la "inteligencia"
del agente vive en esa tabla, no en el ambiente.


### 💬 Actividad 4 — Leer una fila de Q

Como ejemplo tomamos el **estado inicial $s=0$** (esquina superior izquierda).
Tras entrenar (`Q_trained`), sus cuatro valores corresponden a Left/Down/Right/Up.
En el mapa `4x4` determinístico, el camino óptimo desde la salida avanza hacia
la meta por abajo/derecha, de modo que:

- La acción con mayor $Q(0,a)$ es la que inicia ese camino óptimo (típicamente
  **Down** o **Right**, según los desempates del entrenamiento).
- $\arg\max_a Q(0,a)$ devuelve justamente esa acción.

**Interpretación:** la fila $Q(s,\cdot)$ ordena las acciones por su valor a
largo plazo desde $s$; la mayor es la que el agente cree que lo acerca más a la
recompensa final (+1 al llegar a la meta) evitando los huecos (reward 0 y fin
de episodio). *(Ejecuta `Q_trained[0]` para ver los valores numéricos exactos
de esta corrida.)*


## 5. Evaluar la política aprendida


```python
mean_reward, std_reward = evaluate_q_policy(
    env,
    Q_trained,
    n_episodes=100
)

print(f"Mean reward: {mean_reward:.3f}")
print(f"Std reward : {std_reward:.3f}")
```

### 💬 Actividad 5 — Exploration vs. exploitation

**Conclusión.** Durante el **entrenamiento** usamos ε-greedy porque el agente
necesita **explorar**: probar acciones aparentemente subóptimas para descubrir
el ambiente y no quedarse atrapado en una política mediocre por falta de datos.

Durante la **evaluación** ya no queremos aprender, sino **medir qué tan buena
es la política aprendida**. Por eso actuamos de forma puramente greedy
($a=\arg\max_a Q(s,a)$): si siguiéramos explorando, inyectaríamos acciones
aleatorias que ensuciarían la medición y bajarían artificialmente la
recompensa. Explorar sirve para *aprender*; explotar sirve para *usar y
evaluar* lo aprendido.


## 6. Experimento: FrozenLake estocástico


```python
slippery_env = gym.make(
    "FrozenLake-v1",
    map_name="4x4",
    is_slippery=True,
    render_mode="rgb_array"
)

Q_slippery = initialize_q_table(
    slippery_env.observation_space.n,
    slippery_env.action_space.n
)

Q_slippery, rewards_slippery = train_q_learning(
    slippery_env,
    Q_slippery,
    n_episodes=20000,
    learning_rate=0.7,
    gamma=0.95,
    max_epsilon=1.0,
    min_epsilon=0.05,
    decay_rate=0.0005,
    max_steps=100
)

mean_reward, std_reward = evaluate_q_policy(
    slippery_env,
    Q_slippery,
    n_episodes=500
)

print(f"Mean reward: {mean_reward:.3f}")
print(f"Std reward : {std_reward:.3f}")
```

### 💬 Actividad 6 — Determinístico vs. estocástico

**Discusión:**

- **Ambiente:** con `is_slippery=False` la transición es **determinística**: la
  acción elegida siempre mueve al agente en esa dirección. Con
  `is_slippery=True` el hielo es **resbaloso/estocástico**: la acción elegida
  solo se cumple con cierta probabilidad y el agente puede deslizarse hacia una
  dirección perpendicular. Esto hace la tarea mucho más difícil (y la
  recompensa media evaluada baja respecto al caso determinístico).
- **Algoritmo:** la **ecuación de Q-Learning no cambia** —sigue siendo
  $Q(s,a)\leftarrow Q(s,a)+\alpha[r+\gamma\max_{a'}Q(s',a')-Q(s,a)]$—. Lo que
  cambia es que ahora $(r,s')$ es una **muestra de una transición aleatoria**;
  justamente por eso Q-Learning funciona: al promediar muchas experiencias
  mediante el factor $\alpha$, estima el valor esperado sin conocer las
  probabilidades de transición. En entornos estocásticos conviene más
  experiencia (más episodios) y a veces un $\alpha$ más pequeño para que el
  promedio sea estable.


# Cierre de clase

**1. ¿Qué almacena $Q(s,a)$?**

> El retorno esperado (recompensa futura acumulada y descontada por $\gamma$)
> de tomar la acción $a$ en el estado $s$ y seguir después la política
> aprendida. Es el "valor a largo plazo" del par estado-acción.

**2. ¿De dónde sale $\max_{a'}Q(s',a')$?**

> Del propio Q-Learning: una vez observado el siguiente estado $s'$, miramos en
> la Q-table la mejor acción posible desde $s'$. Es la estimación *bootstrap*
> del valor futuro que se usa para construir el TD target
> $r+\gamma\max_{a'}Q(s',a')$.

**3. ¿Por qué necesitamos $\epsilon$-greedy?**

> Para equilibrar **exploración y explotación**. Si siempre explotáramos el
> `argmax`, el agente nunca probaría acciones nuevas y podría quedar atrapado
> en una política subóptima; ε-greedy fuerza a explorar una fracción del tiempo
> y así descubrir mejores caminos.

**4. ¿Por qué Q-Learning es model-free?**

> Porque **no necesita conocer** las probabilidades de transición $T(s,a,s')$
> ni la función de recompensa $R$. Aprende directamente de muestras
> $(s,a,r,s')$ obtenidas al interactuar con el ambiente, sin construir ni usar
> un modelo de su dinámica.
